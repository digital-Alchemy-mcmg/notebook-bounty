## Locate and read the synthetic examples

The cell below works when the notebook is launched from the repository root, as it is in the supplied Codespace.

```python
SOURCE_DIR = Path("synthetic_sources")
source_paths = sorted(path for path in SOURCE_DIR.iterdir() if path.is_file())
assert source_paths, f"No source files found in {SOURCE_DIR.resolve()}"

sources = {path.as_posix(): path.read_text(encoding="utf-8") for path in source_paths}
print(f"Loaded {len(sources)} source documents:")
for source_file, text in sources.items():
    print(f"- {source_file}: {len(text)} characters")
```

## Inspect, validate, and save results

The following cell checks only the public output contract. It does not evaluate extraction quality.

```python
required_fields = {
    "builder_id", "source_file", "source_locator",
    "source_quote", "evidence_blurb"
}
for index, item in enumerate(results):
    missing = required_fields - set(item)
    assert not missing, f"Result {index} is missing fields: {sorted(missing)}"
    assert item["builder_id"] == BUILDER_ID, f"Result {index} has a different builder_id"
    assert item["source_file"] in sources, f"Result {index} names an unknown source"

output_path = Path("output/results.json")
output_path.parent.mkdir(parents=True, exist_ok=True)
output_path.write_text(json.dumps(results, indent=2, ensure_ascii=False) + "\n", encoding="utf-8")
print(f"Saved {len(results)} results to {output_path}")
results[:3]
```
